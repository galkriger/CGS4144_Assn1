```python
import pandas as pd

# Load the original data
df = pd.read_csv("Data/SRP064298/SRP064298.tsv",sep='\t')
df
```

```python
import json
import requests
import time

ensembl_list = list(df["Gene"])


server = "https://rest.ensembl.org"
ext = "/lookup/id"
headers={"Content-Type" : "application/json", "Accept" : "application/json"}

result = {}
for i in range(0, len(ensembl_list), 1000): # Limited because request size limited
    print(i)
    data = '{ "ids" : '+json.dumps(ensembl_list[i:i+min(1000, len(ensembl_list)-i-1)])+' }'
    r = requests.post(server+ext, headers=headers, data=data)
     
    while not r.ok:
        print("ERROR, sleeping 10 seconds and trying again")
        time.sleep(10)
        r = requests.post(server+ext, headers=headers, data=data)
    
    result = result | r.json()
```

```python
# Missed last gene, need to fetch it
r = requests.get(server+ext+"/"+ensembl_list[-1], headers={"Content-Type" : "application/json"})
result = result | {ensembl_list[-1]: r.json()}
gene_names.append(r.json()["display_name"])
```
