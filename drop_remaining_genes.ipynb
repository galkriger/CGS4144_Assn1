```python
import pandas as pd

# Not all gene names were fetched, ~3k were left over.
# Since these didn't have any names in the ensembl database,
# We drop them. We can't do enrichment analysis on them anyways.

df = pd.read_csv("Data/SRP064298/gene_data.tsv",sep='\t')
df
```

```python
unfetched_genes = []

for i, gene in enumerate(df["Gene"]):
    if gene.startswith("ENS"):
        unfetched_genes.append(i)
```

```python
print(len(df["Gene"][unfetched_genes]))
print(df["Gene"][unfetched_genes])
```

```python
df.drop(unfetched_genes).to_csv("Data/SRP064298/gene_data.tsv", sep="\t", index=False)
```

```python
# For unnormalized dataset, drop unknown genes

df2 = pd.read_csv("Data/SRP064298/unnormalized.tsv",sep='\t')
df2 = df2.drop(unfetched_genes)
df2["Gene"] = df["Gene"]
df2.to_csv("Data/SRP064298/unnormalized.tsv", sep="\t", index=False)
```
