```python
import pandas as pd
import sklearn
import numpy as np
import scipy
from sklearn.preprocessing import StandardScaler
```

```python
# Load the original data
df = pd.read_csv("Data/SRP064298/gene_data.tsv",sep='\t')
for col in df.columns:
    if col == "Gene":
        continue
    if int(col[3:]) >= 2534867 and int(col[3:]) <= 2534951:
        df = df.drop(columns=[col])
df
```

```python
# Each gene is a row, so I calculate the variance of each row, sort, then pick the top 5k.
var = []
i = 0
for index, row in df.iterrows():
    var.append(np.var(list(row)[1:]))
print(len(var))
```

```python
df["Variance"] = var
df
```

```python
df_sorted = df.sort_values(by='Variance', ascending=False)
df_sorted = df_sorted.drop(columns=["Variance"])
df_sorted.to_csv("results/assignment3/affinity/sorted_genes.csv",index=False)
df_sorted
```

```python
df_scaled = pd.DataFrame(StandardScaler().fit_transform(df_sorted.drop(columns=["Gene"])))
df_scaled
```

```python
# Method 1: Affinity Propagation
def run_affn_prop(top_n, preference):
    return sklearn.cluster.AffinityPropagation(random_state=42, preference=preference).fit(df_scaled[:top_n].T)

print("Top 5000:")
print("No Preference, Num Clusters:",len(run_affn_prop(5000, preference=None).cluster_centers_))
```

```python
print("Top 5000:")
preferences = [None, -100, -500, -1000, -5000, -10000, -20000]
cluster_results_preference = []
for p in preferences:
    cluster_results_preference.append(run_affn_prop(5000, p))
    print("Preference:", p, "Num Clusters:", len(cluster_results_preference[-1].cluster_centers_))
```

```python
print("Default Preference")
gene_counts = [10, 100, 1000, 5000, 10000]
cluster_results_gene = []

for g in gene_counts:
    clusters = run_affn_prop(g, None)
    cluster_results_gene.append(clusters)
    print("Gene Count:", g, "Num Clusters:", len(clusters.cluster_centers_))
```

```python
for i in range(len(gene_counts)-1):
    for j in range(i+1, len(gene_counts)):
        stat, p_val, _, _ = scipy.stats.chi2_contingency(pd.crosstab(cluster_results_gene[i].labels_, cluster_results_gene[j].labels_))
        print(gene_counts[i], "vs", gene_counts[j], "-", p_val)
```

```python
df_affn_clusters = {"Sample": df_sorted.drop(columns=["Gene"]).columns}

for g in range(len(gene_counts)):
    df_affn_clusters[str(gene_counts[g])+"_genes_No_Preference"] = cluster_results_gene[g].labels_

for p in range(len(preferences)):
    df_affn_clusters["5000_genes_"+str(preferences[p])+"_Preference"] = cluster_results_preference[p].labels_

final_results = pd.DataFrame(df_affn_clusters)
final_results.to_csv("results/assignment3/affinity/cluster_results.csv",index=False)
final_results
```
