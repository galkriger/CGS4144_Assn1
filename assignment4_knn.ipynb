```python
# K-Nearest Neighbors Implementation
import pandas as pd
import sklearn
import numpy as np
import scipy
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_predict, StratifiedKFold
from sklearn.metrics import roc_auc_score
from sklearn.inspection import permutation_importance
from sklearn.pipeline import make_pipeline
```

```python
df = pd.read_csv("results/assignment3/affinity/sorted_genes.csv")
df
```

```python
# Make a list for the experiment labels
kmeans_results = pd.read_csv('results/assignment3/kmeans/cluster_labels.csv')
kmeans_results = kmeans_results.iloc[:, 1:]
sample_groups = []
for sample in kmeans_results["sample_id"]:
    sample_groups.append(int(sample[3:]) <= 2534782) # if this conditional is true, the sample is in the AL group, else RF group
```

```python
kmeans_cluster_groups = kmeans_results["kmeans_g5000_k4"]
kmeans_cluster_groups
```

```python
X = df.drop(columns='Gene')[:5000].T.to_numpy() # Becomes a 2d array for input
clf = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=4))

# getting labels and probabilities from KNN
cv = StratifiedKFold(5, shuffle=True, random_state=42)
labels = cross_val_predict(clf, X, sample_groups, cv=cv)
probs  = cross_val_predict(clf, X, sample_groups, cv=cv, method="predict_proba")
roc_auc_score(sample_groups, probs[:, 1])
```

```python
labels_cluster = cross_val_predict(clf, X, kmeans_cluster_groups, cv=cv)
probs_cluster = cross_val_predict(clf, X, kmeans_cluster_groups, cv=cv, method="predict_proba")
roc_auc_score(kmeans_cluster_groups, probs_cluster, multi_class="ovr")
```

```python
# how gene count affects ROC
gene_counts = [10, 100, 1000, 10000]
print("Experimental Labels:")
for gc in gene_counts:
    X = df.drop(columns='Gene')[:gc].T.to_numpy()
    probs  = cross_val_predict(clf, X, sample_groups, cv=cv, method="predict_proba")
    print("Gene Count:", gc, "| AUC:", roc_auc_score(sample_groups, probs[:, 1]))

print("Cluster Labels")
for gc in gene_counts:
    X = df.drop(columns='Gene')[:gc].T.to_numpy()
    probs_cluster = cross_val_predict(clf, X, kmeans_cluster_groups, cv=cv, method="predict_proba")
    print("Gene Count:", gc, "| AUC:", roc_auc_score(kmeans_cluster_groups, probs_cluster, multi_class="ovr"))
```

```python
X = df.drop(columns='Gene')[:20].T.to_numpy() # Need to set to 20 genes. Any greater yielded all 0's in the importance.
clf.fit(X, sample_groups)
importance = permutation_importance(clf, X, sample_groups, n_repeats=3, random_state=42, n_jobs=-1, scoring="roc_auc")
```

```python
top_10_genes = np.argsort(importance.importances_mean, key=abs)[::-1][:10]
df["Gene"][top_10_genes]
```
